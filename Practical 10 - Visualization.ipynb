```python
# Practical 10: Visualization using Seaborn

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)

df = iris.frame

df.columns = ["sepal_length", "sepal_width",
              "petal_length", "petal_width", "species"]

df["species"] = df["species"].map(dict(enumerate(iris.target_names)))

print(df.head())
```

```python
# Problem 1: Create a Scatter Plot

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="petal_length",
    y="petal_width",
    hue="species"
)

plt.title("Scatter Plot of Iris Dataset")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")

plt.show()

# Analysis:
# The scatter plot shows the relationship between petal length and petal width.
# Different species form different groups.
# Setosa has smaller petals, while Virginica has larger petals.
# Petal length and petal width have a positive relationship.
```

```python
# Problem 2: Create a Density Plot

plt.figure(figsize=(8, 5))

sns.kdeplot(
    data=df,
    x="petal_length",
    hue="species",
    fill=True,
    common_norm=False,
    alpha=0.3
)

plt.title("Density Plot of Iris Petal Length")
plt.xlabel("Petal Length")
plt.ylabel("Density")

plt.show()

# Analysis:
# The density plot shows the distribution of petal length.
# Setosa has a lower petal length.
# Versicolor has a medium petal length.
# Virginica generally has a higher petal length.
```

```python
# Problem 3: Create a Violin Plot

plt.figure(figsize=(8, 5))

sns.violinplot(
    data=df,
    x="species",
    y="petal_length"
)

plt.title("Violin Plot of Iris Dataset")
plt.xlabel("Species")
plt.ylabel("Petal Length")

plt.show()

# Analysis:
# The violin plot shows the distribution of petal length for each species.
# Setosa has a smaller petal length.
# Versicolor has a medium petal length.
# Virginica has a larger petal length.
# The width of the violin represents data density.
```

```python
# Problem 4: Create a Heatmap

plt.figure(figsize=(8, 6))

correlation = df.select_dtypes(include="number").corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Heatmap of Iris Dataset")

plt.show()

# Analysis:
# The heatmap shows the correlation between numerical columns.
# Petal length and petal width have a strong positive correlation.
# Sepal length also has a positive correlation with petal length.
# Sepal width has a negative correlation with petal length.
# Correlation values close to +1 show a strong positive relationship.
# Values close to -1 show a strong negative relationship.
# Values close to 0 show a weak linear relationship.
```

```python
# Problem 5: Create a Joint Plot

sns.jointplot(
    data=df,
    x="petal_length",
    y="petal_width",
    hue="species",
    height=7
)

plt.show()

# Analysis:
# The joint plot shows the relationship between petal length and petal width.
# The central graph displays the scatter plot.
# The top and right sides show the individual distributions.
# Different species form separate groups.
# Petal length and petal width show a positive relationship.
```

```python
# Problem 6: Create a Pair Plot

sns.pairplot(
    data=df,
    hue="species",
    diag_kind="hist"
)

plt.show()

# Analysis:
# The pair plot shows relationships between all numerical features.
# It displays scatter plots between different columns.
# Diagonal graphs show the distribution of individual features.
# Petal length and petal width help distinguish the three species.
# Setosa is clearly separated from the other species.
```
