```python
# 1. Create an empty Dictionary

inventory = {}

print(inventory)
```

```python
# 2. Store the first product details in variables:
# Product Name = Mobile Phone
# Product Quantity = 5
# Product Price = 20000
# Product Release Year = 2020

ProductNo1_Name = "Mobile Phone"
ProductNo1_Quantity = 5
ProductNo1_Price = 20000
ProductNo1_releaseYear = 2020

print("Product Name:", ProductNo1_Name)
print("Product Quantity:", ProductNo1_Quantity)
print("Product Price:", ProductNo1_Price)
print("Product Release Year:", ProductNo1_releaseYear)
```

```python
# 3. Add the first product details in the inventory

inventory["ProductNo1"] = {
    "Name": ProductNo1_Name,
    "Quantity": ProductNo1_Quantity,
    "Price": ProductNo1_Price,
    "ReleaseYear": ProductNo1_releaseYear
}

print("Inventory:", inventory)
```

```python
# 4. Store the second product details in variables

ProductNo2_Name = "Laptop"
ProductNo2_Quantity = 2
ProductNo2_Price = 50000
ProductNo2_releaseYear = 2022

print("Product Name:", ProductNo2_Name)
print("Product Quantity:", ProductNo2_Quantity)
print("Product Price:", ProductNo2_Price)
print("Product Release Year:", ProductNo2_releaseYear)
```

```python
# 5. Add the second product details in the inventory

inventory["ProductNo2"] = {
    "Name": ProductNo2_Name,
    "Quantity": ProductNo2_Quantity,
    "Price": ProductNo2_Price,
    "ReleaseYear": ProductNo2_releaseYear
}

print("Updated Inventory:", inventory)
```

```python
# 6. Display the products present in the inventory

print("Products in Inventory:")

for product, details in inventory.items():
    print(product, ":", details)
```

```python
# 7. Check if ProductNo1_releaseYear and ProductNo2_releaseYear are in the inventory

if inventory["ProductNo1"]["ReleaseYear"] == ProductNo1_releaseYear:
    print("ProductNo1_releaseYear is present in inventory.")

if inventory["ProductNo2"]["ReleaseYear"] == ProductNo2_releaseYear:
    print("ProductNo2_releaseYear is present in inventory.")
```

```python
# 8. Delete the release year of both products from the inventory

del inventory["ProductNo1"]["ReleaseYear"]
del inventory["ProductNo2"]["ReleaseYear"]

print("Inventory after deleting release years:")
print(inventory)
```

```python
# Part B - Tuples
# 1. Create a tuple called prices containing the given values

prices = (250, 300, 150, 400, 100, 350, 200)

print("Prices:", prices)
```

```python
# 2. Find and print the highest and lowest price

print("Highest Price:", max(prices))
print("Lowest Price:", min(prices))
```

```python
# 3. Calculate and print the total sum of all prices

total = sum(prices)

print("Total Sum:", total)
```

```python
# 4. Convert the tuple into a sorted list in ascending order

sorted_prices = sorted(prices)

print("Sorted List:", sorted_prices)
```
