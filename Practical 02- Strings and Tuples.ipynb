```python
# 1. Create an empty list

shopping_list = []

print(shopping_list)
```

```python
# 2. Store the items in the shopping list: Watch, Laptop, Shoes, Pen, Clothes

shopping_list = ["Watch", "Laptop", "Shoes", "Pen", "Clothes"]

print("Shopping List:", shopping_list)
print("Number of items:", len(shopping_list))
```

```python
# 3. Add new item Football to the shopping list

shopping_list.append("Football")

print("Updated Shopping List:", shopping_list)
```

```python
# 4. Print the last item from the shopping list

print("Last Item:", shopping_list[-1])
```

```python
# 5. Print the entire shopping list

print("Entire Shopping List:", shopping_list)
```

```python
# 6. Print the important items like Laptop and Shoes from the shopping list

print("Important Items:")
print(shopping_list[1])
print(shopping_list[2])
```

```python
# 7. Change the item: Instead of Pen, now I want to buy Notebook

shopping_list[3] = "Notebook"

print("Updated Shopping List:", shopping_list)
```
