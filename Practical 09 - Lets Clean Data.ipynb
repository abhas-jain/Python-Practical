```python
# Problem 1: Zipper File

filename = "output.txt"
strings = ["Hello", "World", "Python", "Programming", "Data", "Science"]

with open(filename, "w") as file:

    for i in range(0, len(strings), 2):
        zipped_string = strings[i] + strings[i + 1]

        file.write(zipped_string + "\n")

print("File created successfully!")

with open(filename, "r") as file:
    print(file.read())
```

```python
# Problem 2: Reformat Date

dates = ["01/07/2026", "02/08/2026", "03/03/2026",
         "04/12/2026", "10/05/2026"]

months = {
    "01": "January",
    "02": "February",
    "03": "March",
    "04": "April",
    "05": "May",
    "06": "June",
    "07": "July",
    "08": "August",
    "09": "September",
    "10": "October",
    "11": "November",
    "12": "December"
}

for date in dates:
    
    month, day, year = date.split("/")

    print("Office Hours on", months[month], day + ",", year)
```
