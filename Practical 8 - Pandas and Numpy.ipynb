```python
# Problem 1: Create a DataFrame to display the result

import pandas as pd
import numpy as np

data = {
    "Student": ["David", "Samuel", "Terry", "Evan"],
    "Age": [27, 24, 22, 32],
    "Country": ["UK", "Canada", "China", "USA"],
    "Course": ["Python", "Data Structures", "Machine Learning", "Web Development"],
    "Marks": [85, 72, 89, 76]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Problem 2: Retrieve the Marks column and assign it to a variable

marks = df["Marks"]

print("Marks:")
print(marks)
```

```python
# Problem 3: Retrieve the Country and Course columns and assign them to a variable

country_course = df[["Country", "Course"]]

print(country_course)
```

```python
# Problem 4: Data Loading and Exploration

customer_data = {
    "Customer_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Aarav", "Priya", "Rahul", "Sneha", "Arjun",
             "Neha", "Rohan", "Ananya", "Vikram", "Isha"],
    "Age": [25, 32, 28, 35, np.nan, 42, 29, 31, 45, 27],
    "Gender": ["Male", "Female", "Male", "Female", "Male",
               "Female", "Male", "Female", "Male", "Female"],
    "Purchase_Amount": [1500, 2500, 1800, 3200, 2100,
                        np.nan, 2800, 1900, 3500, 2200]
}

customers = pd.DataFrame(customer_data)

print("First 5 Rows:")
print(customers.head())

print("\nSummary Statistics:")
print(customers.describe())

print("\nMissing Values:")
print(customers.isnull().sum())

customers["Age"] = customers["Age"].fillna(customers["Age"].median())
customers["Purchase_Amount"] = customers["Purchase_Amount"].fillna(
    customers["Purchase_Amount"].median()
)

print("\nDataset After Handling Missing Values:")
print(customers)

print("\nMissing Values After Cleaning:")
print(customers.isnull().sum())
```

```python
# Problem 5: Data Filtering and Selection

print("Customers Older Than 30:")
older_customers = customers[customers["Age"] > 30]
print(older_customers)

male_total = customers.loc[
    customers["Gender"] == "Male", "Purchase_Amount"
].sum()

print("\nTotal Purchase Amount for Male Customers:", male_total)


female_total = customers.loc[
    customers["Gender"] == "Female", "Purchase_Amount"
].sum()

print("Total Purchase Amount for Female Customers:", female_total)

selected_customers = customers[
    ["Customer_ID", "Name", "Purchase_Amount"]
]

print("\nSelected Customer Details:")
print(selected_customers)
```

```python
# Problem 6: Data Transformation and Aggregation

customers["Discounted_Price"] = customers["Purchase_Amount"] * 0.90

print("Dataset with Discounted Prices:")
print(customers)

average_purchase = customers.groupby("Gender")["Purchase_Amount"].mean()

print("\nAverage Purchase Amount by Gender:")
print(average_purchase)

sorted_customers = customers.sort_values(
    by="Purchase_Amount", ascending=False
)

print("\nCustomers Sorted by Purchase Amount (Descending):")
print(sorted_customers)
```
