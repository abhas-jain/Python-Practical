```python
# Q1. Check if Lionel Messi has more than 10 achievements.
# If true, print the player's name, sport, and achievements.
# Otherwise, print that the player does not have more than 10 achievements.

player_name = "Lionel Messi"
sport = "Soccer"
achievements = 7

if achievements > 10:
    print("Player Name:", player_name)
    print("Sport:", sport)
    print("Achievements:", achievements)
else:
    print(player_name, "does not have more than 10 achievements.")
```

```python
# Q3. Check if a player has less than 10 achievements
# and does not play Soccer.
# Print their details if they meet the criteria.

player_name = "Usain Bolt"
sport = "Athletics"
achievements = 8

if achievements < 10 and sport != "Soccer":
    print("Player Name:", player_name)
    print("Sport:", sport)
    print("Achievements:", achievements)
else:
    print("The player does not meet the criteria.")
```

```python
# Q4. Take an integer input from 1 to 9999 and determine:
# 1. Whether the number is even or odd.
# 2. Whether the number is a palindrome.
# 3. Whether the number is divisible by the sum of its digits.

number = int(input("Enter an integer between 1 and 9999: "))

# Check whether the number is even or odd
if number % 2 == 0:
    print("The number is even.")
else:
    print("The number is odd.")

# Check whether the number is a palindrome
if str(number) == str(number)[::-1]:
    print("The number is a palindrome.")
else:
    print("The number is not a palindrome.")

# Calculate the sum of digits
digit_sum = sum(int(digit) for digit in str(number))

# Check divisibility by the sum of digits
if number % digit_sum == 0:
    print("The number is divisible by the sum of its digits.")
else:
    print("The number is not divisible by the sum of its digits.")
```

```python
# Q7. Use a while loop to copy the strings 'orange' from the list
# squares to the list new_squares.
# Stop and exit the loop if the value in the list is not 'orange'.

squares = ['red', 'yellow', 'green', 'purple', 'blue']
new_squares = []

i = 0

while i < len(squares):
    if squares[i] != 'orange':
        break

    new_squares.append(squares[i])
    i += 1

print("New Squares:", new_squares)
```

```python
# Q8. Find the animals whose names are made of 7 letters
# using a while loop and create a separate list of such animals.

Animals = ["lion", "giraffe", "gorilla", "parrots", "crocodile", "deer", "swan"]

seven_letter_animals = []

i = 0

while i < len(Animals):
    if len(Animals[i]) == 7:
        seven_letter_animals.append(Animals[i])

    i += 1

print("Animals with 7 letters:", seven_letter_animals)
```
