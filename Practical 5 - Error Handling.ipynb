```python
# Exercise 1: Create a safe_divide function.


def safe_divide(numerator, denominator):
    try:
        result = numerator / denominator
        return result
    except ZeroDivisionError:
        print("Error: Cannot divide by Zero.")
        return None


numerator = float(input("Enter the numerator: "))
denominator = float(input("Enter the denominator: "))

result = safe_divide(numerator, denominator)

if result is not None:
    print("Result:", result)
```

```python
# Exercise 2: Create a function that calculates the square root

import math

def calculate_square_root(number):
    if number < 0:
        print("Invalid input! Please enter a positive integer or a float.")
        return None
    else:
        return math.sqrt(number)


number = float(input("Enter a positive number: "))

result = calculate_square_root(number)

if result is not None:
    print("Square root:", result)
```

```python
# Exercise 3: Create a function that divides the value of "num"


def perform_calculation(num):
    try:
        result = num / (num - 5)
        print("Result:", result)
    except Exception as e:
        print("An error occurred during calculation.")


num = float(input("Enter the value of num: "))

perform_calculation(num)
```
