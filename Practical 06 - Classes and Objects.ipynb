```python
# Problem 1: Fix the above buggy code using super() in Python.


class Person:

   
    def __init__(self, name, id):
        self.name = name
        self.id = id

   
    def Display(self):
        print("Name:", self.name)
        print("ID:", self.id)


class Emp(Person):

    
    def __init__(self, name, id):
        super().__init__(name, id)

    
    def Print(self):
        print("Emp class called")


Emp_details = Emp("Abhas", 10)


Emp_details.Display()
Emp_details.Print()
```

```python
# Problem 2: Write a Python program to find Anagram Check.


word1 = input("Enter first word: ")
word2 = input("Enter second word: ")

if sorted(word1.lower()) == sorted(word2.lower()):
    print("The strings are anagrams.")
else:
    print("The strings are not anagrams.")
```

```python
# Problem 3: Write a Python program to check for Armstrong Number.


number = int(input("Enter a number: "))

digits = str(number)
power = len(digits)

sum_of_powers = 0

for digit in digits:
    sum_of_powers += int(digit) ** power

if sum_of_powers == number:
    print(number, "is an Armstrong number.")
else:
    print(number, "is not an Armstrong number.")
```

```python
# Problem 4: Create a Vehicle class for a car dealership's inventory management system.

# Task 1: Create a Vehicle class with attributes for maximum speed and mileage.
# Task 2: Set the default color of all vehicles as "white".

class Vehicle:

    def __init__(self, max_speed, mileage):
        self.max_speed = max_speed
        self.mileage = mileage
        self.color = "white"
        self.seating_capacity = 0

    # Task 3: Create a method to assign seating capacity to a vehicle.
    def assign_seating_capacity(self, capacity):
        self.seating_capacity = capacity

    # Task 4: Create a method to display all the properties of a vehicle.
    def display_properties(self):
        print("Maximum Speed:", self.max_speed, "kmph")
        print("Mileage:", self.mileage, "kmpl")
        print("Color:", self.color)
        print("Seating Capacity:", self.seating_capacity)


# Task 5: Create two objects of the Vehicle class.


car1 = Vehicle(200, 20)
car1.assign_seating_capacity(5)

car2 = Vehicle(180, 25)
car2.assign_seating_capacity(4)


# Display the properties of the first vehicle.
print("First Vehicle:")
car1.display_properties()

print()

# Display the properties of the second vehicle.
print("Second Vehicle:")
car2.display_properties()
```
